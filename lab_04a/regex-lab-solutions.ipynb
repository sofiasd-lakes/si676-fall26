## 1. What function do you use in Python to create a regex pattern?

The function `re.compile()` creates a regex pattern.

## 2. What does the re .search() method return?

The method `re.search()` scans a string and returns the first `Match` object that it finds. If it can't find a match, then it returns `None`.

## 3. How can you display or return the strings that are matched?

You can call `.group()` on a `Match` object. Here are four options:

1. `match.group()` or `match.group(0)` returns the whole match.
2. `match.group(1)` returns the text captured by the first capturing group.
3. `match.group(2)` returns the second capturing group (and the third capturing group for (3), etc.).
4. `match.groups()` returns all captured groups as a tuple.

## 4. The findall() method can return a list of strings or a list of tuples. Why would it return one or the other?

The returned result depends on the number of capturing groups in the regex.

1. No capturing groups: a list containing each complete match
2. One capturing group: a list containing that group's text for each match
3. Multiple capturing groups: a list of tuples containing the groups from each match

A noncapturing group, written `(?:...)`, groups parts of a regex without changing the structure of the `findall()` result.

## 5. Find IP addresses in `mbox-short.txt`

The following code reads `mbox-short.txt`, finds IP addresses with `finditer()`, and writes each matched address to `ip-addresses.txt`.

```python
from pathlib import Path
import re

input_path = Path("mbox-short.txt")
output_path = Path("ip-addresses.txt")

ip_pattern = re.compile(r"\b(?:\d{1,3}\.){3}\d{1,3}\b")

with input_path.open("r", encoding="utf-8") as input_file, \
     output_path.open("w", encoding="utf-8") as output_file:

    for line in input_file:
        for match in ip_pattern.finditer(line):
            output_file.write(match.group(0) + "\n")

print(f"Wrote IP addresses to {output_path}")
```
