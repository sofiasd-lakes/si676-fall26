```python
import xml.etree.ElementTree as ET
```

```python
from pathlib import Path

DATA = Path("..","xml-exercise/book-sample.xml")

EAD_sample = DATA / 'book-sample-in-class.xml'

EAD_relpath = 'xml-exercise/book-sample.xml'
```

```python
print(DATA)
```

```python
# parse XML = load + process

tree = ET.parse(DATA)
root = tree.getroot()
```

```python
print(root)
```

```python
print(ET.tostring(root))
```

```python
for element in root:
        print(element)
```

```python
for element in root: 
    print(element.tag, ' --- ', element.attrib, ' --- ',element.text)
```

```python
for element in root.iter():
    print(element.tag, ' --- ', element.attrib)
```

```python
for element in root.iter():
    print(element.tag, ':', element.text)
```

```python
book = tree.find('{http://www.nogood.com/Book}book')
print(book.tag, book.attrib, book.text)
```

```python
for element in book:
    print(element.tag, element.text)
```
