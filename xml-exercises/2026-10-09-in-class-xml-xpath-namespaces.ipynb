XML & XPath & Namespaces

```python
from pathlib import Path

DATA = Path('sample-ead-superior-full.xml')
OUT = Path('..', 'output')

EAD_sample = DATA / 'sample-ead-superior-full.xml'

import xml.etree.ElementTree as ET

tree = ET.parse(DATA)
root = tree.getroot()
```

```python
ns = {
    'ead': 'http://ead3.archivists.org/schema/',
    'dcterms': 'https://purl.org/???'
}
```

```python
control = root.find('{http://ead3.archivists.org/schema/}control')

print(control.tag, ' -- ', control.attrib)
```

```python
control2 = root.find('ead:control', ns)
print(control2.tag)
```

Using XPath in .findall()

```python
#scopecontent
for note in tree.findall('//ead:scopecontent', ns):
    print(note.tag)
```

XPath to look for components (<c>)

```python
c_count = 0

for component in tree.findall('.//ead:c', ns):
    c_count += 1
    print(component.tag, ':', component.attrib)

print(f'Found {c_count} component(s).')
```

```python
c_count = 0

for component in tree.findall('.//ead:c[@level="series"]', ns):
    c_count += 1
    print(component.tag, ':', component.attrib)

print(f'Found {c_count} component(s).')
```

```python
c_count = 0

for component in tree.findall('.//ead:c[@level="series"]', ns):
    c_count += 1
    print(component.tag, ':', component.text)

print(f'Found {c_count} component(s).')
```
